## 3. PMMの基本形（概要）
PMMは**主行列**と**副行列**から出力を作る。

- 主行列: $P_j(
{
{c_l}
})$（入力特徴 $
{
{c_l}
}$ に依存するHermitian/Unitary行列）
- 固有ベクトル $v_j^{(i)}$ を使って出力を構成
- 副行列: $S_k$ を用いた双線形形式

典型的なスカラー出力（イメージ）

$$
y = \left| \left(v^{(i)}
\right)^\dagger S_k v^{(j)} 
\right|^2
$$

量子力学の**期待値・遷移確率**に似た形になる点が特徴。

## 4. AE-PMM と AO-PMM（2種類の基本形）
### 4.1 AE-PMM（Affine Eigenvalue PMM）
入力に対し主行列が**一次（アフィン）**に変化: 
$$
M(c)=nderline{M_0}+um_{l=1}^p c_lnderline{M_l}
$$
出力は $M(c)$ の固有値。

### 4.2 AO-PMM（Affine Observable PMM）
固有ベクトルと副行列の双線形で**複数出力**を作る。
回帰・分類に便利（softmaxと組み合わせ可能）。
